```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = {
    'Name': ['A', 'B', 'C', 'D', 'E'],
    'Math':[65, 89, 67, 65, 45],
    'Science':[66, 99, 62, 85, 83],
    'English': [85, 79, 92, 65, 88]
}
df = pd.DataFrame(data)
print(df)
```

```python

marks = df[['Math', 'Science', 'English']].values

average_marks = np.mean(marks, axis=1)

df['Average'] = average_marks

print(df)
```

```python
def grade(avg):
    if avg >= 90:
        return 'A'
    elif avg >= 75:
        return 'B'
    elif avg >= 60:
        return 'C'
    elif avg >= 50:
        return 'D'
    else:
        return 'F'

df['Grade']=df['Average'].apply(grade)
df
```

```python
import matplotlib.pyplot as plt

# Data
subjects = ["Python", "SQL", "English", "Computer"]
average_marks = [85, 78, 88, 92]

# Plotting bar chart
plt.figure()
plt.bar(subjects, average_marks)
plt.xlabel("Subjects")
plt.ylabel("Average Marks")
plt.title("Average Marks of Students")
plt.show()
```
